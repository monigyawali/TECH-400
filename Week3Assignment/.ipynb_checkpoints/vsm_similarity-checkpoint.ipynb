## Required Libraries ##

```python
import os

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

import pandas as pd
import matplotlib.pyplot as plt
```

## Loading the Documents ##

```python
DOCS_FOLDER = "."

def load_documents(folder):
    documents = {}

    for filename in sorted(os.listdir(folder)):
        if filename.endswith(".txt"):
            path = os.path.join(folder, filename)

            with open(path, "r", encoding="utf-8", errors="ignore") as f:
                documents[filename] = f.read()

    return documents


documents = load_documents(DOCS_FOLDER)

print("Loaded documents:", list(documents.keys()))
print("Number of documents:", len(documents))
```

## Building the TF-IDF Vector Space Model ##

```python
filenames = list(documents.keys())
corpus = list(documents.values())

vectorizer = TfidfVectorizer(stop_words="english")

tfidf_matrix = vectorizer.fit_transform(corpus)

print("TF-IDF matrix shape:", tfidf_matrix.shape)
print("Sample vocabulary terms:")
print(vectorizer.get_feature_names_out()[:15])
```

## Calculating Cosine Similarity ##

```python
similarity_matrix = cosine_similarity(tfidf_matrix)

sim_df = pd.DataFrame(
    similarity_matrix,
    index=filenames,
    columns=filenames
)

print("Cosine Similarity Matrix:\n")
print(sim_df.round(3))
```

## Ranking the Document Pairs ##

```python
pairs = []

for i in range(len(filenames)):
    for j in range(i + 1, len(filenames)):
        pairs.append(
            (
                filenames[i],
                filenames[j],
                similarity_matrix[i][j]
            )
        )

pairs.sort(key=lambda x: -x[2])

print("All pairwise similarities (highest to lowest):\n")

for a, b, score in pairs:
    print(f"{a} <-> {b}: {score:.3f}")
```

## Visualizing Similarity Using a Heatmap ##

```python
fig, ax = plt.subplots(figsize=(9, 7))
im = ax.imshow(sim_df.values, cmap="Blues", vmin=0, vmax=1)

ax.set_xticks(range(len(sim_df.columns)))
ax.set_yticks(range(len(sim_df.index)))
ax.set_xticklabels(sim_df.columns, rotation=45, ha="right", fontsize=8)
ax.set_yticklabels(sim_df.index, fontsize=8)

for i in range(len(sim_df.index)):
    for j in range(len(sim_df.columns)):
        ax.text(j, i, f"{sim_df.values[i, j]:.2f}", ha="center", va="center", color="black", fontsize=7)

plt.title("Document Similarity Heatmap (Cosine Similarity)")
plt.colorbar(im)
plt.tight_layout()
plt.savefig("similarity_heatmap.png", dpi=150)
plt.show()
```
